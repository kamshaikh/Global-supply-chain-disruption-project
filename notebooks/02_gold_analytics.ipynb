```python
import sys
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))
from config.pipeline_config import get_spark, silver_path, gold_path, bronze_path
spark = get_spark("notebook")
```

```python
from pyspark.sql import functions as F
risk = spark.read.format('delta').load(gold_path('gold_supplier_risk'))
risk.orderBy(F.desc('risk_score')).show(10, False)
```

```python
risk.groupBy('risk_tier').agg(F.count('*').alias('suppliers'), F.round(F.avg('risk_score'),1).alias('avg_risk')).show()
```

```python
spark.read.format('delta').load(gold_path('gold_inventory_risk')).filter("priority = 'CRITICAL'").show(20, False)
```
